# Data Cleaning & Preparation

### Checking the number of missing values in each column


```python
db.isna().sum()
```

### Missing Value Analysis

The dataset contains missing values in six columns:

* Director, Cast, and Country is strings and contain a significant number of missing values. Removing these rows would result in the loss of a large amount of valuable data, so these missing values will be replaced with "Unknown".

* rating i will Replace missing ratings with "Unavailable"

* Date Added is mixed-type values so, instead of removing missing values. i will seperate it to 2 other coilmns but aslo filtered out the rows with Nan values into 0. 

* some 'rating' values accidentally contain duration values (e.g., "74 min"). so, We move misplaced duration values from 'rating' to 'duration'



### Replacing (director, Cast, and Country) missing values with "Unknown"

```python
db["director"]= db["director"].fillna("unknown")
db["cast"]=db["cast"].fillna("unknown")
db["country"]= db["country"].fillna("unknown")
```

```python
# checking if missing values filled succsesfuly
db.isna().sum()
```

### Replacing missing ratings with "Unavailable"


```python
db['rating'] = db['rating'].fillna("Unavailable")
```

```python
# checking if missing values filled succsesfuly
db.isna().sum()
```

### now for Date_added i will seperate it into 2 coulmns (month added , year added) and replace missings with 0 

```python
# Extract the month word and 4-digit year directly using regular expressions
db["month_added"] = db["date_added"].str.extract(r"([A-Za-z]+)")
db["year_added"] = db["date_added"].str.extract(r"(\d{4})")
```

```python
db["year_added"]=db["year_added"].fillna(0)
db["month_added"]=db["month_added"].fillna(0)
db["date_added"]=db["date_added"].fillna(0)

# checking if missing values filled succsesfuly
db.isna().sum()
```

### Fix Mixed-Type & Column Inconsistencies ('duration' & 'rating')

```python
# In the Netflix dataset, some 'rating' values accidentally contain duration values (e.g., "74 min")
# We move misplaced duration values from 'rating' to 'duration'

misplaced_mask = db['rating'].str.contains('min|Seasons', na=False)

db.loc[misplaced_mask, 'duration'] = db.loc[misplaced_mask, 'rating']
db.loc[misplaced_mask, 'rating'] = "Unavailable"

# Extract numerical duration and unit type into separate columns
db['duration_num'] = db['duration'].str.extract(r'(\d+)').astype(float)
db['duration_type'] = db['duration'].str.extract(r'([a-zA-Z]+)')

# Normalize unit names ('min' -> 'Minutes', 'Season'/'Seasons' -> 'Seasons')
db['duration_type'] = db['duration_type'].replace({
    'min': 'Minutes',
    'Season': 'Seasons',
    'Seasons': 'Seasons'
})
```

```python
# checking the new columns
db.head()
```

### Parse Dates & Create New Time Columns

```python
# Clean date string formatting
clean_dates = db['date_added'].str.strip()

# Parse into proper pandas datetime objects
# Handles variations with or without spaces after the comma (e.g., "September 24, 2022")
db['date_added_dt'] = pd.to_datetime(
    clean_dates, format='%B %d, %Y', errors='coerce'
).fillna(
    pd.to_datetime(clean_dates, format='%B %d,%Y', errors='coerce')
)

# Extract new required feature columns
db['year_added'] = db['date_added_dt'].dt.year.astype('Int64')
db['month_added'] = db['date_added_dt'].dt.month_name()
db['month_added_num'] = db['date_added_dt'].dt.month.astype('Int64')
```

```python
# Drop temporary helper column
db = db.drop(columns=['date_added_dt'])
```
